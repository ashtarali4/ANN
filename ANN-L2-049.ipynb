```python
import numpy as np
```

## Task 1: Vectors

```python
# Vector of ages for 6 people
ages = np.array([25, 30, 45, 52, 38, 60])

print("Ages vector:", ages)
print("Shape:", ages.shape)
print("Dimensions (ndim):", ages.ndim)
print("Data type:", ages.dtype)
```

```python
# Years until retirement (65 - age), computed without a loop (vectorized operation)
years_until_retirement = 65 - ages
print("Years until retirement:", years_until_retirement)
```

```python
# Average age using a NumPy function (not manual calculation)
average_age = np.mean(ages)
print("Average age:", average_age)
```

## Task 2: Matrices

```python
# 4x3 matrix: rows = products, columns = months
sales = np.array([
    [120, 150, 130],   # Product 1
    [80,  90,  100],   # Product 2
    [200, 210, 190],   # Product 3
    [60,  75,  70]     # Product 4
])

print("Sales matrix:\n", sales)
print("Shape:", sales.shape)
print("Total number of elements:", sales.size)
```

```python
# Same-shaped matrix filled with zeros using a NumPy function
zero_matrix = np.zeros(sales.shape)
print("Zero matrix:\n", zero_matrix)
```

```python
# Replace the second row (index 1) using indexing, not recreation
zero_matrix[1] = [10, 20, 30]
print("Updated matrix:\n", zero_matrix)
```

## Task 3: Addition & Subtraction

```python
# Rows = students, Columns = subjects
term1 = np.array([
    [70, 65, 80],   # Student 1
    [55, 60, 58],   # Student 2
    [90, 85, 88]    # Student 3
])

term2 = np.array([
    [75, 70, 85],   # Student 1
    [65, 72, 60],   # Student 2
    [92, 80, 95]    # Student 3
])

print("Term 1 marks:\n", term1)
print("Term 2 marks:\n", term2)
```

```python
# Total combined marks for each student in each subject
total_marks = term1 + term2
print("Total combined marks:\n", total_marks)
```

```python
# Improvement (Term2 - Term1) for each student in each subject
improvement = term2 - term1
print("Improvement matrix:\n", improvement)
```

```python
# Identify the student with the highest total improvement using code
student_improvement_totals = improvement.sum(axis=1)   # sum across subjects for each student
best_student_index = np.argmax(student_improvement_totals)

print("Total improvement per student:", student_improvement_totals)
print(f"Student with highest improvement: Student {best_student_index + 1} "
      f"(total improvement = {student_improvement_totals[best_student_index]})")
```

## Task 4: Dot Product

```python
# Prices of 4 items and quantities purchased
prices = np.array([50, 20, 15, 100])
quantities = np.array([2, 5, 3, 1])

# Total bill using dot product (not manual multiply+add)
total_bill = np.dot(prices, quantities)
print("Total bill:", total_bill)
```

```python
# 2x4 matrix: rows = customers, columns = quantities of each of the 4 items
customers_quantities = np.array([
    [2, 5, 3, 1],   # Customer 1
    [1, 2, 4, 3]    # Customer 2
])

# Total bill for both customers using a single dot product operation
total_bills = customers_quantities.dot(prices)
print("Total bill for each customer:", total_bills)
```

## Task 5: Reshaping

```python
# 1D array of hours studied over 12 days
hours_studied = np.array([2, 3, 1, 4, 2, 5, 3, 1, 2, 4, 3, 2])
print("Original 1D array:", hours_studied)
print("Original shape:", hours_studied.shape)
```

```python
# Reshape into 4 rows x 3 columns (4 weeks of a 3-day study plan)
reshaped_4x3 = hours_studied.reshape(4, 3)
print("Reshaped to 4x3 (4 weeks x 3 days):\n", reshaped_4x3)
```

```python
# Reshape into a different valid shape, e.g. 3x4 (3 weeks x 4 days)
reshaped_3x4 = hours_studied.reshape(3, 4)
print("Reshaped to 3x4 (3 weeks x 4 days):\n", reshaped_3x4)

print("\nBoth reshaped versions:")
print("4x3 version:\n", reshaped_4x3)
print("3x4 version:\n", reshaped_3x4)
```

```python
# Try an invalid reshape (5x3 = 15 elements, but we only have 12) to see the error
try:
    invalid_reshape = hours_studied.reshape(5, 3)
except ValueError as e:
    print("Error message received:")
    print(e)

# Note: The error occurs because 5 x 3 = 15, but the original array only has
# 12 elements. NumPy requires the total number of elements to stay the same
# before and after reshaping.
```

## Task 6: Indexing

## Task 7: Slicing

```python
# 6x6 matrix with numbers 1 to 36 in order
matrix_6x6 = np.arange(1, 37).reshape(6, 6)
print("6x6 matrix:\n", matrix_6x6)
```

```python
# Top-left 3x3 sub-matrix
top_left_3x3 = matrix_6x6[0:3, 0:3]
print("Top-left 3x3 sub-matrix:\n", top_left_3x3)
```

```python
# Every alternate row and every alternate column (step slicing)
alternate_rows_cols = matrix_6x6[::2, ::2]
print("Every alternate row and column:\n", alternate_rows_cols)
```

```python
# Last 2 rows and last 2 columns together as a single sub-matrix
last_2x2 = matrix_6x6[-2:, -2:]
print("Last 2 rows & last 2 columns sub-matrix:\n", last_2x2)
```

## Task 9: Transpose

```python
# 3x5 matrix: rows = employees, columns = monthly performance scores
performance = np.array([
    [85, 90, 88, 92, 95],   # Employee 1
    [70, 75, 78, 80, 82],   # Employee 2
    [60, 65, 63, 70, 72]    # Employee 3
])

print("Original performance matrix:\n", performance)
print("Original shape:", performance.shape)
```

```python
# Transpose the matrix
performance_T = performance.T
print("Transposed matrix:\n", performance_T)
print("New shape:", performance_T.shape)
```

```python
# Verify programmatically that transposing twice returns the original matrix
double_transpose = performance.T.T
is_same = np.array_equal(performance, double_transpose)
print("Does double transpose equal the original matrix?", is_same)
```

```python
# Create a 4x2 and a 3x2 matrix
matrix_4x2 = np.array([
    [1, 2],
    [3, 4],
    [5, 6],
    [7, 8]
])

matrix_3x2 = np.array([
    [1, 0],
    [0, 1],
    [1, 1]
])

print("4x2 matrix:\n", matrix_4x2)
print("3x2 matrix:\n", matrix_3x2)

result = matrix_4x2.dot(matrix_3x2.T)
print("Dot product result (4x2 . (3x2)^T = 4x3):\n", result)
print("Result shape:", result.shape)
```
